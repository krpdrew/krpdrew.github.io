##### (Optional) Example: Vectorization

Python's built-in `math` module already handles basic mathematical operations seamlessly, so it’s fair to ask: what does `NumPy` actually bring to the table?

The answer lies in how Python handles data.

The native `math` module is designed to work on one number at a time. What about 1000?
We will encounter loops soon, but even they are notoriously inefficient in standard Python.

This is where `NumPy` solves the problem through **vectorization**, allowing us to run calculations across entire numeric arrays almost instantly.

Consider the code below, where we want to **find the square roots of the first million whole numbers.**

We provide runtimes with both `math` and `NumPy`.

````python
import math
import numpy as np

# Create the dataset
data_size = 1000000
python_list = list(range(data_size))
numpy_array = np.arange(data_size)

print("Standard Python (math.sqrt + 'for' loop) runtime:")
%timeit [math.sqrt(x) for x in python_list]

print("\nNumPy (np.sqrt) runtime:")
%timeit np.sqrt(numpy_array)

print(np.sqrt(numpy_array))

# WARNING! Attempting to print python_list or the loop output might cause a crash
# However, NumPy arrays are safe to print since text is automatically truncated!
````

The result makes it clear that for large datasets common in data analysis, scientific simulations, and machine learning, vectorization is significantly faster and more efficient than standard iteration through arrays.
