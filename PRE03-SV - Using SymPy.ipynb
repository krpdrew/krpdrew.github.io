## Section D: Taylor series in sympy

A Taylor expansion approximates a smooth function near a specific point (called the 'base point') using an infinite sum of derivatives.

The Taylor expansion of a function $ U(x) $ about the base point $ x = x_0 $ is given by:

$$ U(x) \sim U(x_0) + U'(x_0)(x-x_0) + \frac{U''(x_0)}{2!}(x-x_0)^2 + \dots $$

**Syntax:**

* We can perform Taylor expansions with `SymPy` using the function `sp.series`

* There are two equivalent ways:

  * Method 1:
    ```
    taylor_expansion = sp.series(function, variable, base_point, n)
    ```
  * Method 2:
    ```
    taylor_expansion = function.series(variable, base_point, n)
    ```
  💡 Inputting `n` leads to all non-zero terms up to (but not including) $\mathcal{O}(x^n)$ being displayed.

* By default, `sp.series` retains a leading error term, represented as $\mathcal{O}(x^n)$. Remove this with:
  ```
  taylor_expansion.removeO()
  ```
    💡 (Capital letter `O`, not the number `0`)

### D.1 Example

````python
x = sp.symbols('x')
expr = sp.sin(x)

display(expr.series(x,0,4))
display(sp.series(expr,x,0,4).removeO())

# Note that since we wanted the cubic (n=3) term, we needed to use 4 while calling the function
````
