The interaction energy between two bonded atoms is modelled using various analytical functions. In this assignment, you will analyze two: the Morse and the Lennard-Jones potentials.

1. **Morse potential**:

  $$ U_{Mo}(r) = D(e^{-2a(r-r_{0})} -2e^{-a(r-r_0)}),$$

  where $r$ is the inter-atomic distance and $D$ (dissociation energy), $r_0$ (equilibrium bond length), $a$ (stiffness parameter) are positive constants.

2. **Lennard-Jones potential**:

  $$ U_{lj}(r) = 4 \epsilon \left(\left(\frac{\sigma}{r_{lj}}\right)^{12} - \left(\frac{\sigma}{r_{lj}}\right)^6 \right),$$

  where $r$ is the inter-atomic distance and $\epsilon$ (dissociation energy/well depth), $\sigma$ (equilibrium bond length) are positive constants.
