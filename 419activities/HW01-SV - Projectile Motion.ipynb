## Section A: Run simulation

We use a numerical simulation that calculates $x(t)$, $y(t)$, $v_x(t)$, and $v_y(t)$ for a projectile with air resistance linear in velocity.

After we find $x(t)$ and $y(t)$, we will use the simulation and the numerical solution to find the range of the projectile.

The simulation relies on Newton's 2nd law to calculate $v(t+dt)$ and $x(t+dt)$ iteratively using the Euler method (ref. PRE01 and PRE02)

```
    # Update velocity using Euler's method
    vx[i] = vx[i-1] + ax * dt
    vy[i] = vy[i-1] + ay * dt
    # Update position using Euler's method
    x[i] = x[i-1] + vx[i] * dt
    y[i] = y[i-1] + vy[i] * dt
```

The condition `if y[i] < 0:` detects when the projectile hits the ground ($y = 0$). At this time `t_stop`, the simulation saves the final trajectory arrays (`x_stop`, `y_stop`) and stores the impact position in `land_x`.

---

👉 Simply run the code below, but pay close attention to the algorithm. You'll need it for future sections.

````python
# Import libraries

import numpy as np
import matplotlib.pyplot as plt

# Define constants
g = 9.81  # Acceleration due to gravity (m/s^2)

# Define initial conditions
v0 = 50.0  # Initial velocity (m/s)
angle = 40
theta = np.radians(angle)  # Launch angle (radians)
x0 = 0.0  # Initial x position (m)
y0 = 0.0  # Initial y position (m)

# Air resistance parameters (simplified)
C = 0.1  # Modified drag coefficient (kg/s)

# Time step and total time
dt = 0.01  # Time step (s)
t_max = 30.0  # Maximum simulation time (s)
t = np.arange(0, t_max, dt)

# Initialize arrays for position and velocity
x = np.zeros_like(t)
y = np.zeros_like(t)
vx = np.zeros_like(t)
vy = np.zeros_like(t)

# Set initial position and velocity
x[0] = x0
y[0] = y0
vx[0] = v0 * np.cos(theta)
vy[0] = v0 * np.sin(theta)

# Simulate projectile motion using Euler's method

for i in range(1, len(t)):

    # Calculate the current velocity magnitude
    v = np.sqrt(vx[i-1]**2 + vy[i-1]**2)

    # Calculate the air resistance force components (proportional to velocity)
    F_drag_x = -C * vx[i-1] #choose correct equation
    F_drag_y = -C * vy[i-1] #choose correct equation  C=b/m =1/tau

    # Calculate the acceleration components
    ax = F_drag_x
    ay = -g + F_drag_y  # Gravity acts downwards

    # Update velocity using Euler's method
    vx[i] = vx[i-1] + ax * dt
    vy[i] = vy[i-1] + ay * dt

    # Update position using Euler's method
    x[i] = x[i-1] + vx[i] * dt
    y[i] = y[i-1] + vy[i] * dt

    '''Below is the essential code for determining the range.'''

    # Stop the simulation if the projectile hits the ground
    if y[i] < 0:
        # oops we went too far, better stop the simulation
        land_time = t[i-1]
        land_x = x[i-1]

        # Truncate the arrays to the landing point
        t_stop = t[:i]
        x_stop = x[:i]
        y_stop = y[:i]
        vx_stop = vx[:i]
        vy_stop = vy[:i]

        break # Exit the loop
````
