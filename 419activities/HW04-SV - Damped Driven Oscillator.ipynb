## Section 0: Damped Drive Oscillator Background Theory

### (0.1) Equation of motion

[References: Chapter 5, *Classical Mechanics* by Taylor]

The equation of motion of a mass experiencing a combination of restoring, damping, and driving forces is given by

$$ m \frac{d^{2}x}{dt^{2}}+b\frac{dx}{dt} + kx = F \cos(ω_{drive}t),$$

where $x(t)$ denotes the position of the mass and the constants are $m$ (mass), $b$ (damping coefficient), $k$ (spring constant), $F$ (driving amplitude), and $\omega_{drive}$ (driving frequency).

### (0.2) Solutions

Define:
*   Natural frequency: $\omega_0 \equiv \sqrt{k/m}$
*   Damping attenuation parameter: $\beta = \frac{b}{2m}$
*   Driving force amplitude per unit mass: $f_0 = \frac{A}{m}$

There are four key dynamic regimes. In each, the solution can be described as $$x(t) = x_h (t) + x_p (t),$$
representing the transient and steady-state solutions respectively.

---

#### Steady-state solution (all regimes):

\begin{align}
x_p (t) &= A(\omega_{drive}) \cos(\omega_{drive}t - \delta),\\
\text{ where }\;\; A(\omega_{drive}) &\equiv \frac{f_0}{\sqrt{(\omega_0^2 - \omega_{drive}^2)^2 \, + \, 4 \, \beta^2 \omega_{drive}^2}}  \\
\delta &\equiv \arctan \left(\frac{2 \,\beta \,\omega_{drive}}{\omega_0^2 \, - \, \omega_{drive}^2}\right)
\end{align}

---

#### Transient solutions:

1.   Undamped ($\beta = 0$)

$$ x_h(t) = A_1 \cos (\omega_0 t) + A_2 \sin (\omega_0 t) $$

2.   Underdamped ($\beta < \omega_0$)

$$ x_h(t) = A_1 e^{-\beta t} \cos \Big(\sqrt{\omega_0^2 - \beta^2} \, t - \delta_1 \Big) $$

3.   Critically damped ($\beta = \omega_0$)

$$ x_h(t) = (A_1 + A_2 t) e^{-\beta t} $$

4. Overdamped ($\beta > \omega_0$)

$$ x_h(t) = A_1 e^{-\big(\beta - \sqrt{\beta^2 - \omega_0^2}\big)t} + A_2 e^{-\big(\beta + \sqrt{\beta^2 - \omega_0^2}\big)t} $$

Note: $A(\omega_{drive})$ is a function. $A_1$, $A_2$, or $\delta$. The latter are arbtitrary constants, determined by the initial conditions $x(0)$ and $v(0)$.

```python
# (0) Import libraries
import matplotlib.pyplot as plt
import numpy as np
from ipywidgets import interact, FloatSlider # this is for fun sliders of parameters

# (i) Define  fixed variables

dt = 0.01  # Time step (s)
t_max = 100.0  # Maximum simulation time (s)
t = np.arange(0, t_max, dt) # Time array
w0 = 2 # Natural frequency w0^2 = sqrt(k/m) (rad/s)

# The following constants are technically determined by x0 and v0
delta = 0 # Phase
A1 = 1 # Amplitude 1
A2 = 1 # Amplitude 2

# (ii) Define the plotting function that will be controlled by sliders

def plot_oscillations(beta, w_drive, F):

    # Undamped
    if beta == 0:
      decay_osc = A1 * np.cos(w0*t) + A2 * np.sin(w0*t)

    # Underdamped
    if beta < w0:
      decay_osc = A1 * np.exp(-beta * t) * np.cos(np.sqrt(w0**2 - beta**2) * t - delta)

    # Critically damped
    elif beta == w0 and beta != 0:
      decay_osc = (A1+A2*t)*np.exp(-beta*t)

    # Overdamped
    else:
      decay_osc = A1*np.exp(-(beta-np.sqrt(beta**2 - w0**2))*t)+A2*np.exp(-(beta + np.sqrt(beta**2-w0**2))*t)

    # Driven oscillation
    driven_osc = (F/np.sqrt((w0**2 - w_drive**2)**2 + 4 * w_drive**2 * beta**2))*np.cos(w_drive * t - np.arctan2(2 * w_drive * beta , (w0**2 - w_drive**2)))

    total_osc = decay_osc + driven_osc

    # Creating plots

    plt.figure(figsize=(10, 14)) # this is a nice way to combine plots into a single "figure"

    plt.subplot(3, 1, 1)
    plt.plot(t, decay_osc)
    plt.xlabel('time (s)')
    plt.ylabel('position (m)')
    plt.title('Transient solution')

    plt.subplot(3, 1, 2)
    plt.plot(t, driven_osc)
    plt.xlabel('time (s)')
    plt.ylabel('position (m)')
    plt.title('Steady-state solution')

    plt.subplot(3, 1, 3)
    plt.plot(t, total_osc)
    # plt.plot(t, decay_osc)
    # plt.plot(t, driven_osc)
    plt.xlabel('time (s)')
    plt.ylabel('position (m)')
    plt.title('Total solution')

# (iii) Create interactive sliders for beta, w_drive, and F

interact(plot_oscillations,
         beta=FloatSlider(min=0.01, max=2.0, step=0.01, value=0.2, description='Damping (b):'),
         w_drive=FloatSlider(min=0.1, max=3.0, step=0.1, value=5.0, description='w_drive:'),
         F=FloatSlider(min=0., max=1.0, step=0.1, value=5.0, description='Driving Amplitude'))
```
