## Appendix: Spring (Hooke's Law) *[PRE03]*

### Plot

The potential energy of a mass on a spring is:

$$ U = \frac{1}{2} k (x - x_0)^2 $$

We will assume:
* $ k = 2 \text{ N/m}$
* $ x_0 = 0.5 \text{ m}$

```python
# Import libraries
import sympy as sp

# Define symbolic variables
# k  = separation
# x  = length scale
# x0 = equilibrium position
k, x , x0 = sp.symbols('k  x x0')

# Write and display equation for potential energy

U = 0.5*k*((x-x0)**2)
display(U)

# Write and display an equation for U with specific values substituted
specificU=U.subs({x0: 0.5, k: 2})
display(specificU)

# plot - check and make sure you picked a good range for x and y
sp.plot(
    U.subs({x0: 0.5, k: 2}),
    (x, -1, 2),
    ylim=(-2, 2),
    xlabel='x',
    ylabel='U(x)',
    title='U(x) for spring potential'
)
```

### Find extrema

Find the minimum of the potential by taking the derivative.

The syntax to take a derivative is

    # Taking a derivative
    U_nth_derivative=sp.diff(U,x,n)

    #showing the equation
    display(U_nth_derivative)

How do you then solve for the position at which the derivative is zero? sympy has a ``sp.solve`` function that allows you to solve for the solution of an equation. The syntax is below. If you want the solution to be a specific number, remember to use the ``.subs`` call

```python
# Take derivatives
''' First derivative '''
U_prime=sp.diff(U,x)
display(U_prime)

'''Second derivative'''
U_prime_prime=sp.diff(U,x,2)
display(U_prime_prime)

# Find all the equilibrium points (it could be more than 1)
equilibrium_points = sp.solve(U_prime, x)
display(equilibrium_points)

# The code here displays just one equilibrium point (the 1st one at spot [0] in the array)
second_deriv_at_equilibrium=U_prime_prime.subs({x:equilibrium_points[0]})
display(second_deriv_at_equilibrium)

# Is/are the equilibrium point(s) a min or max?
#  Lets look at all the equilibrium points
# (In this example there is only one, but there could be many)

for equilibrium_point in equilibrium_points:
  second_deriv_at_equilibrium=U_prime_prime.subs({x:equilibrium_point})
  if second_deriv_at_equilibrium.subs({k:2}) >0 :
    print (f"at x={equilibrium_point} this is a min")
  else :
    print (f"at x={equilibrium_point} this is a max")
```
