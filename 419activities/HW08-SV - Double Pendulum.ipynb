```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, HTMLWriter
from IPython.display import HTML

plt.close('all')

# Get values and compute coordinates
m1_val, m2_val, L1_val, L2_val = 1.0, 1.0, 0.5, 1.0
theta1, theta2, time, L1, L2 = doublependulum(m1_val, m2_val, L1_val, L2_val)

x1 = L1 * np.sin(theta1)
y1 = -L1 * np.cos(theta1)
x2 = x1 + L2 * np.sin(theta2)
y2 = y1 - L2 * np.cos(theta2)

# Speed up animation rendering (take every 10th point)
step = 10
t_anim = time[::step]
x1_anim = x1[::step]
y1_anim = y1[::step]
x2_anim = x2[::step]
y2_anim = y2[::step]

# Set up figure and animation elements
fig, ax = plt.subplots(figsize=(5, 5))
ax.set_xlim(-(L1 + L2 + 0.2), L1 + L2 + 0.2)
ax.set_ylim(-(L1 + L2 + 0.2), L1 + L2 + 0.2)
ax.set_aspect('equal')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('Double Pendulum Motion')

line, = ax.plot([], [], 'k-', lw=2)
masses, = ax.plot([], [], marker='o', color='red', linestyle='', markersize=8)

# Initialization function
def init():
    line.set_data([], [])
    masses.set_data([], [])
    return line, masses

# Update function
def update(i):
    # Rod coordinates: Origin -> Mass 1 -> Mass 2
    line.set_data([0, x1_anim[i], x2_anim[i]], [0, y1_anim[i], y2_anim[i]])
    # Mass positions
    masses.set_data([x1_anim[i], x2_anim[i]], [y1_anim[i], y2_anim[i]])
    return line, masses

# Create animation
anim = FuncAnimation(fig, update, init_func=init, frames=len(t_anim), interval=30, blit=True)

plt.close(fig)
HTML(anim.to_jshtml())
```
