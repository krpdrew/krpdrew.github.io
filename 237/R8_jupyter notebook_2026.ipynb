## Background / context

The photoelectric effect demonstrates that light transfers energy in discrete quanta (photons). Electrons are ejected from a metal only when the photon energy exceeds the material's work function, and the emitted electrons' kinetic energy increases linearly with the light frequency above the cutoff frequency.


Photoelectric effect: https://en.wikipedia.org/wiki/Photoelectric_effect
###Physical constants:
* Speed of light: $c = 3.0 × 10^8$ m/s
* Electron rest mass energy: $m_ec^2$ = 511 keV
* h = $4.14 × 10^{−15} $eV · s
* hc = 1240 eV · nm
* The work function $\phi$ for cesium is 1.9 eV, the lowest of any metal.
