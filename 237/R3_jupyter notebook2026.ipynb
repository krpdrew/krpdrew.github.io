## Background / context
In this activity you will use Python as a tool to visualize and compare exact formulas with their Taylor-series approximations. These approximations are widely used in physics to understand complicated equations by considering limits in which they become simpler.

As an example, we'll look at relativistic corrections to physics at high-velocities. These relativistic corrections may seem complicated, but in the limit of low-velocities (meaning $v/c\ll1$) then we expect to find familiar formula with corrections that depend on higher powers of $v/c$. For example, kinetic energy $K \sim \frac{1}{2}mv^2 + \frac{3}{8}mv^4/c^2 + \mathcal{O}(v^6/c^4)$.

## Using an equation 'solver'

To use solve an equation, you can use "solve":

```
solve(equation_to_solve,variable_you_are_solving_for)`
```
importantly the "equation you want to solve" is only one side of the equation (and assumes the other side is zero) .

## Example (y=mx+b):

if your equation is `y=m*x+b`
and you want to find the expression for x when y=5, you would solve for `0=m*x+b-5`. This is shown below:

````python
m,x,b = symbols ('m,x,b')

solve(m*x+b-5,x)
````
