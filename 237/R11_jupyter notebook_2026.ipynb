## Background / context


The wave function describing a state of an electron confined to move along the x axis is given at time zero by:

$\psi(x, 0) = Ae^{− x^{2}/4\sigma^{2}}$

A is a normalization constant we have to calculate such that
$\int^{\infty}_{-\infty}|\psi(x)|^2 dx=1$

in the python code below we have to rearrange this equation to solve for A.

$\frac{1}{\int^{\infty}_{-\infty}|\psi(x)/A|^2 dx}=A^2$

we call $|\psi(x)/A|^2$ = Psi_squared_divided_by_A_squared
