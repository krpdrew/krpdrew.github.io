## Background / context

The infinite square well is one of the simplest exactly solvable systems in quantum mechanics. Confining a particle between infinitely high potential walls forces its wavefunction to form standing waves, which leads directly to quantized energy levels.

You will first explore standing-wave solutions, then calculate their energy levels, and finally simulate a moving wave packet.

````python
# This is a standing wave in an infinite square well (just solutions, no animation)
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

# Define parameters
L = 1  # Length of the box [nm]
x = np.linspace(0, L, 1000)  # x-axis values in nm

def plot_wave_function(n):
    # Calculate the wave function
    psi = np.sqrt(2/L) * np.sin(n * np.pi * x / L)

    # Calculate the probability density
    prob_density = psi**2

    # Plot the wave function and probability density
    plt.figure(figsize=(10, 6))
    plt.plot(x, psi, label=f'Wave function (n={n})')
    plt.plot(x, prob_density, label='Probability density')
    plt.axvline(0, color='k', linestyle='--', label='Well boundary') # Add left wall
    plt.axvline(L, color='k', linestyle='--') # Add right wall
    plt.xlabel('Position (x) [nm]')
    plt.ylabel('Amplitude / Probability Density')
    plt.title(f'Wave Function and Probability Density in an Infinite Square Well (n={n})')
    plt.legend()
    plt.show()

# Create interactive widget for n
n_slider = widgets.IntSlider(min=1, max=10, step=1, value=2, description='Energy Level (n):')

# Link widget to the plotting function
interactive_plot = widgets.interactive(plot_wave_function, n=n_slider)

# Display the widget and the interactive plot
display(interactive_plot)
````

````python
# This is a simulation of a wave packet trapped in a well
import numpy as np
import matplotlib.pyplot as plt
from scipy.linalg import inv
from IPython.display import display, clear_output
import time as t
import ipywidgets as widgets

# Initialize values
m = 1  #  m_e = 9.10938356e-31  # Mass of electron (kg)
hbar = 1     ### in reality : 1.0545718e-34  # Reduced Planck constant (Joule second)
dt = 0.25  # the units of time are approximately 10^-16s
time = np.arange(0, 800 + dt, dt)
nt = len(time)

L = 50 #barrier width # (*10^-10 m approx)
dx = 0.1
L_array= 200 #  Angstroms  #this is the length of the array, not the length of the well
x = np.arange(-L_array/2, L_array/2 + dx, dx)
mx = len(x)


# Initialize potential
v0 = 50  #units of * hbar**2 / (2 * m_e) *k0**2
V = np.zeros(mx)
for j in range(mx):
    if x[j] < -L/2 or x[j] > L/2:
        V[j] = v0

# Initialize Hamiltonian (independent of K)
H = np.zeros((mx, mx), dtype=complex)
for i in range(mx):
    H[i, i] = 2 / dx**2 + V[i] # diagonal

for i in range(mx - 1):
    H[i, i+1] = -1 / dx**2 # off diagonal
    H[i+1, i] = -1 / dx**2

# Initialize Crank-Nicolson matrix (independent of K)
identity_matrix = np.eye(mx)
matrix = inv(identity_matrix + 1j * dt / 2 * H) @ (identity_matrix - 1j * dt / 2 * H)


def run_simulation(K):
    # Initialize wave packet (depends on K)
    x0 = 0
    #wavelength =de Broglie wavelength (meters) h/p approximately
    wavelength=2*np.pi*hbar/np.sqrt(K*2*m)
    sigma = wavelength/4#  Width of the Gaussian packet (meters)

    k0=np.sqrt(K * 2 * m)/hbar # =2*np.pi/lambda
    K_label=K
    Psi = np.zeros((mx, nt), dtype=complex)
    Psi[:, 0] = (2 * np.pi * sigma**2)**(-1/4) * np.exp(-(x - x0)**2 / (2* sigma**2)) * np.exp(1j * k0 * x)

    # Simulation loop
    inttest = np.zeros(nt)
    inttest[0] = np.sum(np.conj(Psi[:, 0]) * Psi[:, 0]) * dx

    # Set up plot
    plt.figure(figsize=(10, 6))

    for t_step in range(nt - 1):
        # Evolve wave function over time
        Psi[:, t_step + 1] = matrix @ Psi[:, t_step]

        # Optional: Plot every 10 time steps
        if t_step % 50 == 0:
            clear_output(wait=True) # Clear the output for animation effect
            plt.plot(x*10, 1000 * np.abs(Psi[:, t_step + 1])**2, label='Probability Density')
            plt.plot(x*10, V, 'r--', label='Potential')
            plt.ylim([-v0, v0 * 1.2]) # Maintain consistent ylim
            plt.xlim([-500, 500]) # Maintain consistent xlim
            plt.xlabel('x (nm)')
            plt.ylabel('Probability Density')
            plt.title(f'Wave Packet Simulation in a Box. \n Time step (time runs to 3000): {t_step} \n Kinetic energy set to {K_label} \n you can type your K-value in the box!')
            plt.legend()
            plt.show()
            t.sleep(0.01) # Pause to control animation speed

        # Test of normalization
        inttest[t_step + 1] = np.sum(np.conj(Psi[:, t_step + 1]) * Psi[:, t_step + 1]) * dx


# Create interactive widget for K
k_slider = widgets.FloatSlider(min=0.001, max=.05, step=0.001, value=0.008, description='Initial Kinetic Energy (K):', readout_format='.4f')

# Link widget to the simulation function
interactive_plot = widgets.interactive(run_simulation, K=k_slider)

# Display the widget and the interactive plot
display(interactive_plot)
````
