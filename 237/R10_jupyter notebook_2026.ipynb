## Create 5 waves

Consider waves of the form: $y=A \sin(kx-\omega t)$.

We want to make a few waves with different $k$'s.

To do this we first import libraries, and then setup the variables:

2.   create an array of k values (k is the wavenumber)

at t=0 this is similar to figure 5.16 in the book.

## Animating waves moving
Now, we are going to look at what happens as a wave packet moves in time. To do this we need a wave-speed and an omega.

Below is the code, with many comments in green to show the steps.

##Q5

**Q5** : Locate the wave number (k_val), omega (w_val), and wave speed used in the code below.

The variable 'k_val' iterates through the array 'k' defined above (find it!).

Write down wave speed (a number) and the equation for w_val on your paper.


````python
# Q5
#  Here we are using for-loops to plot the sum of the sine waves (the wave-packet)
# as it moves through time with velocity 10. We change the equation to be sin(k*x-w*t)

from IPython.display import clear_output # we need this trick to help with visualization (I looked it up)

# Define the time array
t = np.linspace(0, .5, 50) # 50 time steps between 0 and 0.5

# Loop over time --> if you haven't seen loops before this is a way to iterate through
# a variable (in this case its t for time) each time calculating a new value for the position (y_sum) of the wave
for time_val in t:
  # Calculate the sum of sine waves at the current time step
  y_sum = np.zeros_like(x) # first - we create an empty placeholder array of y_sum values

  for k_val in k: # now we sum up all the waves (what are the k-values?)
    w_val = 10* k_val # what is this?
    y_sum =y_sum+ np.sin(k_val * x - w_val * time_val)

  # Redraw the plot
  clear_output(wait=True) # Clear the output of the current graph: super important for visibility
  plt.plot(x,y_sum) # Redraw the figure
  plt.xlabel('x')
  plt.ylabel('y')
  plt.show()
#^^ all of this is indented and therefore changes at each time-step,
````
