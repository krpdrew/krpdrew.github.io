## Writing symbolic equations

Below we define our variables, these are called "symbols" in python.

After we define the "symbols" we can then write an equation using them.

Here are some standard mathematical expressions:

*    a**b is $a^b$
*    a*b is $a\times b$
*    exp(x) is $e^x$
*    sqrt(x) is $\sqrt{x}$
*    derivatives are calculated using  ```diff```, as in: ```diff(equation, variable)```
*    Taylor series can be calculated in python via:
```
taylor_expansion=series(equation_you_are_approximating, variable, value_you_are_approximating_around, # of derivatives+1)
```
Or, you can write
```
taylor_expansion=equation_you_are_approximating.series(variable,value_you_are_approximating_around,# of derivatives+1)
```
When you put this all together, it might look like the following example.
### Example


````python
# Install sympy libraries
## the * means import all the sympy functions
from sympy import *

#  Writing the (symbolic) equation
##  Define the symbols
d, f = symbols('d f')

## Define the equation
equation= exp(d) + sqrt(d**2) + f

print(equation) #this shows the equation below

# Take the first derivative of the equation with respect to d

derivative_of_equation=diff(equation,d)

print(derivative_of_equation)

# Take the second derivative with respect to d

second_derivative_of_equation=diff(equation,d,2)
print(second_derivative_of_equation)

# Take a taylor series expansion with respect to d around 0 to 3rd order, O(d**3)

taylor_equation=equation.series(d,0,3)
print(taylor_equation)

# Note that there is an O(d**3) term -- we want to get rid of that and only look at terms up to d**2

remove_O=taylor_equation.removeO()
print(remove_O)
````

## Graphing functions

**Step 1** Define a function ('newequation')

**Step 2** Plot the function

You can modify the function you are plotting, specify plot ranges etc. The syntax basics are:
```
plot(function,xlim=(a,b),ylim=(c,d), xlabel = "label",ylabel="label")

OR

plot(function1, function2, xlim=(a,b),ylim=(c,d), xlabel = "label", ylabel="label")
```
### Example

````python
# define symbols
x = symbols('x')

# write equation
newequation=1/2*sin(x)**2+cos(x)

# plot  with limits from zero to pi
plot(newequation,xlim=(0,pi))

# add axes labels
plot(newequation,xlim=(0,pi),xlabel="x", ylabel="trig functions")

# you can also set ylim
plot(newequation,xlim=(0,pi),ylim=(0,1), xlabel="x", ylabel="trig functions")
````

## Integrating functions

Sympy only spits out the answer, it does not explain the steps or the substitutions used.

This will never be enough for recitations and homework, but it is useful as a check.
### Example
(Note, Ne(a,b) means $a\neq b$)


````python
# Install sympy libraries
from sympy import *  # the * means import all the sympy algorithms

# Define your symbols
x,A,k = symbols('x A k')

# Integrate x**2
integral=integrate(x**2, x)
print(integral)

# Lets try a definite integral
# It is important to note in sympy that "log" means "natural log" ("ln")
integral=integrate(1/(exp(-2*k*x)), (x, 0, 1))
print(integral)
````
