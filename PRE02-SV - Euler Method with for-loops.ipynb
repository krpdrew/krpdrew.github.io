## Section A: Constant velocity motion

* We are going to practice using a ```for``` loop to create lists of physical values (arrays).

* Below we will show a basic technique for solving for x(t) given a constant velocity, we use the for-loop to integrate over all time points separated by dt.
*   The **Euler method** is a numerical technique that approximates solutions to continuous differential equations.

*   For constant velocity motion, the Euler method takes a pretty simple form:

$$ x(t) = \int v dt \xrightarrow{discretized} x_{i+1} =x_{i}+ (v\cdot dt)$$

*   As you can see, this formula instructs us to calculate the next position ($x_{i+1}$) by taking the current position ($x_i$) and adding the small distance traveled ($v \cdot dt$) during that time.

Now let's try to implement this in code. We will assume $y_0 = 0 \text{ m}$ and $ v = 5 \text{ m/s}$ (again, units don't enter our code).

#### A1 Initializing time and position

````python
#  create an array of time values ################################
time= np.linspace(0, 2, 1001)

#  define the other variables in your equation for y(t) ##########
position = np.zeros_like(time) #this sets up a blank array of positions
                               # position is now a set of [0,0,0,0, etc] that is 1001 numbers long
v=5 # units will be arbitrary - pretend its m/s

# define your initial position and velocity  #################
position[0]=0 # the 0th position spot is our "initial position"

# define your dt
dt = time[1]-time[0] # defining the distance dt between 2 time points in the time array
````

#### A2 create the "for" loop to create an array of position values for each time-point

````python
#       in the for loop code we use an integer 'i' which is incremented from 0 to N-1 where
#       N is the length of the position array (len(position))
for i in range(len(position)-1):
  position[i+1]= position[i]+v*dt #this is "Euler's method but should be familiar as kinematics equation"
# ^^ notice indent!
````
