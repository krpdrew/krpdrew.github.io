## Section B: Run example code

ODE solvers generate a numerical sequence of values. Rather than solving differential equations analytically, they excel at finding solutions iteratively through numerical methods.

Earlier in the semester, we numerically solved ODEs using the Euler method with for-loops; while that approach requires more lines of code, it is much easier to follow step-by-step.

**Our example:**

Consider a one-dimensional first-order ODE:

$$\begin{align}
\quad
\frac{dv}{dt} = -g, \quad \mbox{with} \quad v(0) = 10 \text{ (ignoring units)}
\end{align}$$

In this case, we can find the analytical solution by separation of variables. We arrive at:

* Analytical solution: $v - v_0 = - g (t - 0)$
* w/ initial conditions: $ v(t) = 10 - gt $

Now, we will check this numerically. We will use the `SciPy` function `solve_ivp`, for which we will need to provide:
1. a function defining the derivative `dv_dt`
2. an array of time points `t_pts`
3. initial condition `v_0`

Then, `solve_ivp` will return a set of arguments, with one being an array of solutions i.e. one value of $v(t)$ for each point in `t_pts`.


👉 Ensure you understand the algorithm. Then simply run the code below.

#### B1 Import library and define derivative function

````python
#  (i) Import libraries for derivative function
from scipy.integrate import odeint


# This is a really simple example
def dv_dt(t, v, g=9.8):
    return -g
````

#### B2 initialize variables

````python
# Define time array and initial conditions
t_pts = np.linspace(0., 10., 101)
v_0 = 10.0
````

#### B3 use solve_ivp

the syntax is:

`solve_ivp( function for rhs of ODE equation,time span (start, end),initial value of v(t),array of t values to evaluate at )`

````python
# Do the numerical integration
v_pts = solve_ivp(dv_dt, (t_pts[0], t_pts[-1]), [v_0], t_eval=t_pts)
````

#### B4 Pull out the values of v(t) from the solving

````python
# Extract the velocity solution array
v=v_pts.y[0]
# Get a sense of what the output looks like
print(v)
````

#### B5  Plot v(t)

````python
# Plot
# We plot both the numerical solution 'v' as well as the analytical solution '10 - gt'

plt.plot(t_pts, v, label='numerical solution')
plt.plot(t_pts, 10-9.8*t_pts, label='analytical solution')
plt.xlabel('t [sec]')
plt.ylabel('v(t) [meters/sec]')
plt.legend()
plt.show()
````
