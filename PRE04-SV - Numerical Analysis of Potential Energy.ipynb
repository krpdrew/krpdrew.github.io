*  In particular, we will consider the potential energy of a cube (mass $m$, side $2 b$) rocking without slipping on top of a fixed cylinder (radius $r$).

*  This is given by the function:

$$ U(\theta) = mg[(r+b)\cos\theta + r\theta\sin(\theta)] $$

*  $\theta$ is the angle between the contact point between cube and cylinder and the vertical.

## Section B: Finding Local Extrema with `SciPy` Example

One way to identify local maxima is by using the `find_peaks` function from the `scipy.signal` library.

By simply modifying the function `U` to `-U`, we can find local minima.

👉 Simply run the code below.

````python
# Importing a function from another the library
from scipy.signal import find_peaks

# Find peaks in (-U), which correspond to local minima in U
local_minima_indices, _ = find_peaks(-U)

# Display

print("Local minima are at the following (theta, U) values:")
for index in local_minima_indices:
    print(f"theta = {round(theta[index],3)}, U = {round(U[index],3)}")
````
